## 3. Dataset Loading & Patient 020 Validation
Initialize `Prostate2DDataset` and verify dimensions, binarization, tumor slices, and DataLoader compatibility.

```python
from src.dataset import Prostate2DDataset

# Auto-detect dataset root (Google Colab vs local workspace)
colab_dataset_root = "/content/drive/MyDrive/THESIS_PROSTATE158/dataset/prostate158_train/train"
local_dataset_root = os.path.join(project_root, "dataset", "prostate158_train", "train")
dataset_root = colab_dataset_root if os.path.exists(colab_dataset_root) else local_dataset_root

print(f"Using dataset root: {dataset_root}")

# 1. Initialize dataset with slice_sampling='all'
dataset = Prostate2DDataset(
    dataset_root=dataset_root,
    patient_ids=["020"],
    modalities=["t2", "adc", "dwi"],
    mask_name="t2_tumor_reader1.nii.gz",
    slice_sampling="all"
)

print(f"Total 2D slices (dynamic): {len(dataset)} (Expected: 24)")
assert len(dataset) == 24, f"Expected 24 slices, got {len(dataset)}"

# 2. Check sample 0 dimensions and types
sample0 = dataset[0]
print(f"Sample 0 Image Shape: {sample0['image'].shape} (Expected: (3, 270, 270))")
print(f"Sample 0 Mask Shape:  {sample0['mask'].shape} (Expected: (1, 270, 270))")
print(f"Sample 0 Mask Unique: {np.unique(sample0['mask'])} (Expected: [0.])")
assert sample0['image'].shape == (3, 270, 270)
assert sample0['mask'].shape == (1, 270, 270)

# 3. Check sample 7 (tumor slice) unique values
sample7 = dataset[7]
print(f"Sample 7 Mask Unique: {np.unique(sample7['mask'])} (Expected: [0., 1.])")
assert np.array_equal(np.unique(sample7['mask']), [0.0, 1.0])

# 4. Identify all slices with positive tumor annotations
tumor_slices = [i for i in range(len(dataset)) if np.any(dataset[i]['mask'] > 0)]
print(f"Tumor slice indices: {tumor_slices} (Expected: [5, 6, 7])")
assert tumor_slices == [5, 6, 7], f"Expected [5, 6, 7], got {tumor_slices}"

# 5. Test PyTorch DataLoader integration
try:
    from torch.utils.data import DataLoader
    loader = DataLoader(dataset, batch_size=2, shuffle=False)
    batch = next(iter(loader))
    print(f"DataLoader Batch Images: {batch['image'].shape} (Expected: torch.Size([2, 3, 270, 270]))")
    print(f"DataLoader Batch Masks:  {batch['mask'].shape} (Expected: torch.Size([2, 1, 270, 270]))")
    assert tuple(batch['image'].shape) == (2, 3, 270, 270)
    assert tuple(batch['mask'].shape) == (2, 1, 270, 270)
    print("DataLoader compatibility check PASSED!")
except ImportError:
    print("PyTorch not installed in this environment; skipping DataLoader test.")
```

## 4. Visualization
Inspect axial slice 7 with multi-modal channels and tumor mask overlay.

```python
fig, axes = plt.subplots(1, 4, figsize=(16, 4))
channel_names = ["T2-weighted", "ADC Map", "DWI", "Tumor Mask"]

for c in range(3):
    axes[c].imshow(sample7['image'][c], cmap="gray")
    axes[c].set_title(channel_names[c])
    axes[c].axis("off")

axes[3].imshow(sample7['image'][0], cmap="gray")
mask_overlay = np.ma.masked_where(sample7['mask'][0] == 0, sample7['mask'][0])
axes[3].imshow(mask_overlay, cmap="Reds", alpha=0.7)
axes[3].set_title("T2 + Tumor Mask Overlay")
axes[3].axis("off")

plt.tight_layout()
plt.show()
```
